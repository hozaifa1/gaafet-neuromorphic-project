```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

# Load the data you extracted
experimental_data = pd.read_csv('exp_forward.csv')
simulated_data = pd.read_csv('Id_Vg_forward.csv')
```

```python

# Plot both curves to see the mismatch
plt.figure(figsize=(8, 6))
plt.plot(experimental_data[experimental_data.columns[0]], experimental_data[experimental_data.columns[1]], label='exp forward',linestyle='None', marker='o')
plt.plot(simulated_data[simulated_data.columns[0]], simulated_data[simulated_data.columns[1]], label='simulated forward', marker='o', linestyle='None')

plt.xlabel('Gate Voltage (Vgs)')
plt.ylabel('Drain Current (Ids)')
plt.yscale('log')
plt.title('Comparison of 4.9eV vs. 4.8eV Workfunction')
plt.legend()
plt.grid(True, which="both", ls="--")  
plt.show()
```
